## CPython 如何實作 Reference？

從 CPython 的角度來看，這種 reference 關係是透過 C 語言的指標（Pointer）實作的。

在 CPython 裡的物件是使用 PyObject 這個「結構（Struct）」來表示的，來看看 PyObject 是怎麼定義的。在 Include/pytypedefs.h 檔案裡有寫到`typedef struct _object PyObject;`。順著 _object 繼續往下找，會發現在 Include/object.h 可以看到 `_object` 這個結構的定義。`_object` 結構看起來還算簡單，在這個結構裡有一個 `_PyObject_HEAD_EXTRA` 巨集以及兩個成員變數，分別是 `ob_refcnt` 以及 `ob_type`，從`_PyObject_HEAD_EXTRA` 開始，在同一個檔案裡看到它的定義：
```c
#define _PyObject_HEAD_EXTRA            \
    PyObject *_ob_next;           \
    PyObject *_ob_prev;
```
在 C 語言裡 #define 是用來定義「巨集（Macro）」用的，巨集不是變數也不是函數，它是在進行編譯的過程中會被替換成指定的程式碼片段，主要的目的是讓程式碼更容易閱讀，也能被重複使用。在巨集上面那行註解：`Define pointers to support a doubly-linked list of all live heap objects.`也就是說，其實 `_PyObject_HEAD_EXTRA` 這個巨集就是 `_ob_next` 跟 `_ob_prev` 這兩個指標，是用來串接所有的 Python 物件，形成一個雙向鏈結串列（Doubly-Linked List）的結構。Python 的物件是存放在 Heap 上的，Heap 不像 Stack 是連續的記憶體位置，這些物件宣告之後可能會散落在 Heap 上，這樣 Python 要怎麼知道目前這個物件的下一個或上一個物件是什麼？Python 使用「雙向鏈結串列（Doubly-Linked List）」的資料結構來串接所有的物件，好處是可以容易的找到上一個或下一個物件，而且在新增或刪除物件的時候也很有效率（O(1)），不過缺點就是在搜尋的時候會比較慢（O(n)）。

從原始碼可以看的出來，`_PyObject_HEAD_EXTRA` 這個巨集在一般的編譯過程通常是空值，只有在定義了 `Py_TRACE_REFS` 時，才會包含 `_ob_next` 和 `_ob_prev` 這兩個指標，通常只有在可能想要抓問題或是想要追蹤 Python 物件的引用計數時才會開啟這個功能，我們一般常用的 Python 版本，物件之間並不會形成一個雙向鏈結串列。

在 `PyObject` 的最後一個 `ob_type` 是一個指向 `PyTypeObject` 的指標，這個 `PyTypeObject` 是什麼呢？`typedef struct _typeobject PyTypeObject;`看來 `PyTypeObject` 就是 `_typeobject` 這個結構的別名。這個結構裡面有很多成員變數，像是 `tp_name` 就是這個類別的名字，`tp_dealloc` 是這個類別的解構函數，不同的類別在做解構的時候可能都會有自己要做的事情，另外 `tp_doc` 就是我們在印出物件的 `.__doc__` 屬性的時候會被印出來的傢伙。 Python 中所有東西都是物件（Everything is an object），每個物件都有它的型別，而 `PyTypeObject` 就是這些型別的核心描述結構。它是 Python 的核心結構之一，負責描述所有的 Python 型別（types），負責定義物件的行為、屬性和方法。完整的 `PyTypeObject` 結構的可在 `Include/object.h` 原始碼裡找到。這個結構包含了關於一個型別所需的所有資訊，例如型別的名稱、大小、方法、屬性，以及該型別如何處理記憶體分配、屬性讀取、物件建立和銷毀等。

再繼續往下看看 `PyVarObject` 的定義。這個 `PyVarObject` 裡面有兩個成員變數，其中 `ob_base` 就是一個 `PyObject`，另一個是 `ob_size` 是用來記錄或表示這個物件的「數量」。什麼意思？例如以串列（List）來說，`ob_size` 就是串列裡面的元素個數，對字串來說就是 `ob_size` 就是字串的長度。對於可變動的物件（例如 List），當增加或刪除元素的時候 `ob_size` 也會跟著變動，相對的，對於不可變動的物件，像是字串或是 `Tuple`，`ob_size` 就是固定的，不會變動。更精準的說，不可變物件沒有提供可以修改 `ob_size` 的方法。

`PyObject` 是 `CPython` 裡的物件結構，這個結構體在 Python 的內部運作中扮演著非常重要角色，其中 `ob_refcnt` 是用來計算物件的引用次數，當引用次數為 0 時，這個物件就會被回收。而 `ob_type` 則是指向 `PyTypeObject` 的指標，這個 `PyTypeObject` 是用來描述物件的型別，包含了物件的行為和特性，整個 Python 的物件導向幾乎就是繞著它轉。正是有 `ob_type` 的存在，每個 `PyObject` 才能有不同的型別，能夠實現物件導向的特性，讓物件能夠進行多型（Polymorphism）和繼承（Inheritance）等操作。

## 從記憶體位址理解 C Pointer

從硬體角度來看，Pointer 本質上只是一個儲存記憶體位址的普通數值。
假設有：
```c
int x = 10;
```
記憶體可能長這樣：
```text
Address      Value
1000         10
1004         ?
1008         ?
```
Compiler 決定：
```text
x 位於位址 1000
```
```c
int *p = &x;
```
Compiler 產生的概念其實是：把 1000 存進 p，而記憶體變成：
```text
Address      Value
1000         10      ← x
2000         1000    ← p
```
因此：p的內容其實就是：1000這個數字。

例如：
```c
int x = 10;
int *p = &x;

printf("%d", *p);
```
Compiler 看到：
```c
*p
```
知道意思是：
```text
取出 p 的內容
↓
得到位址 1000
↓
讀取位址 1000 的資料
↓
得到 10
```
概念上：

```text
p
 ↓
1000
 ↓
10
```
**C 的 Pointer 是對「記憶體位址」的抽象；而 Python 的 Reference 則是對「Pointer」再進一步的抽象。**
這也是為什麼能透過 CPython 的 `PyObject *` 來理解 Python 的 Reference，而能透過 Compiler 與 Memory Layout 來理解 C 的 Pointer。

#### 鴨子型別（duck typing）
Python有使用一種（非官方）鴨子型別( duck typing )。這其實不是什麼技術術語，它來自一句老話：

如果它看起來像鴨子、走路像鴨子、叫磬也像鴨子，那很有可能就是鴨子。

在許多程式語言中，系統會特別在意一個物件的型別（Type）是否符合指定的類別。而在 Python 中，程式通常更關心：這個物件能做什麼？
而不是這個物件屬於哪個類別？例如，假設某個物件支援：
- 加法（`+`）
- 減法（`-`）
- 乘法（`*`）
- 除法（`/`）
以及其他數值運算。那麼在許多情況下，Python 並不在乎它究竟是：int、float，甚至是程式設計者自行建立的類別。只要它表現得像數值（Numeric Type），就可以參與數值運算。
 
例如：

```python
x = 100
y = 3.14
print(x + y)
```
Python 不會因為：x 是 int，y 是 float，而拒絕執行。相反地，Python 會檢查這些物件是否支援加法運算。更進一步，我們甚至可以建立自己的類別：
```python
class Temperature:
...
```
只要這個類別實作了適當的運算，它也可能像數值一樣被使用。因此，Duck Typing 的核心思想可以簡單表示為：不要只看物件是什麼（What it is）。更要看物件能做什麼（What it can do）。

輸出結果應為月支付額及總支付額，可藉由以下公式取得：

$$
\textit{monthlyPayment} = \frac{ \textit{loanAmount}\times\textit{monthlyInterestRate} }{ 1-\dfrac{1}{ \left(1+\textit{monthlyInterestRate}\right)^{\textit{numberOfYears}\times 12} } } 
$$

$$
\textit{totalPayment} = \textit{monthlyPayment} \times \textit{numberOfYears} \times 12 
$$

這是房屋貸款（月付款，Monthly Payment）常見的等額本息（amortized loan）公式。
因此，此程式所需的輸入資料有月利率、貸款年數，以及貸款總額。

步驟1. 提示使用者輸入年利率、總年數，以及貸款總額。

步驟2. 年利率的輸入格式為百分比，比方説4.5%。程式必須藉由除以100，將這個數值轉換為十進位數字。由於一年有12個月，因此可藉由將年利率除以12，來取得月利率。綜合上述所説，我們必須將以百分比表示的年利率除以1200，以取得十進位表示的月利率。
例如，假設年利率為4.5％，那麼月利率就會是4.5/1200 = 0.00375 。

步驟3. 利用步驟2 的公式計算月支付額。

步驟5. 顯示月支付額及總支付額。

階段4 實作

實作又被稱作撰寫程式碼（coding）。在前面的公式裡，我們必須計算

$$ (1+\textit{monthlyInterestRate})^{\textit{numberOfYears}\times 12}， $$

您可使用**運算子撰寫如下
```python
(1 + monthlyInterestRate)**(numberOfYears*12)
```

````python
# Computeloan.py

# Enter yearly interest rate
annualInterestRate = eval(input(
  "Enter annual interest rate, e.g., 8.25: "))
monthlyInterestRate = annualInterestRate / 1200


# Enter number of years
numberOfYears = eval(input(
  "Enter number of years as an integer, e.g., 5: "))
    
# Enter loan amount
loanAmount = eval(input("Enter loan amount, e.g., 120000.95: "))
    
# Calculate payment
monthlyPayment = loanAmount * monthlyInterestRate / (1
  - 1 / (1 + monthlyInterestRate) ** (numberOfYears * 12))
totalPayment = monthlyPayment * numberOfYears * 12


# Display results
print("The monthly payment is", int(monthlyPayment * 100) / 100)
print("The total payment is", int(totalPayment * 100) /100)

````

### 3. Python 的 `time` 模組是否自行維護時間？

不是。

`time` 模組本身並不負責維護時間。

它比較像是一個介面（interface），負責向作業系統查詢時間資訊。

```text
Python time 模組
        ↓

        ↓
   系統時間
```

因此：


並不是 Python 自己計算出來的。

而是向作業系統取得目前時間後，再轉換成 Python 物件。

#### RTC（Real-Time Clock）

#### 網路時間伺服器（NTP）

現代作業系統還會定期透過：

```text
NTP
(Network Time Protocol)
```

向網路時間伺服器同步時間。

因此整體流程可表示為：

```text
原子鐘
    ↓
NTP Server
    ↓
作業系統時間
    ↓
Python time 模組
```

### 5. 如果電腦沒有網路，這些函式是否仍能運作？請說明理由。

可以。

例如：

```python
time.time()
time.gmtime()
time.localtime()
```

即使在沒有網路的情況下仍然可以執行。

原因是：

```text
time 模組主要向作業系統查詢時間，
而作業系統通常可以從 RTC
(Real-Time Clock)
取得目前時間。
```

因此：

```text
沒有網路
    ≠
沒有時間資訊
```

不過如果：

- RTC 電池沒電
- 長時間未進行網路校時

則時間可能逐漸產生誤差。

因此現代電腦通常會：

```text
RTC 負責持續計時
+
NTP 負責定期校正
```

共同維持正確時間。

## 參考解答

### 1. 為什麼程式語言喜歡採用共同的時間標準？

共同的時間標準可以提高不同系統之間的相容性（compatibility）。

例如：

- Windows
- Linux
- macOS

都可能需要交換時間資料。

若大家都使用相同的時間表示方式（例如 Unix Timestamp），則可以直接比較與計算時間，而不需要進行額外轉換。

例如：

```python
t1 = 1790499503
t2 = 1790503103

elapsed = t2 - t1
```

即可得到：

```text
3600 秒
```

若每個作業系統使用不同標準，則交換資料前必須先進行轉換，增加程式設計的複雜度。

因此：

> 使用共同的時間標準有助於跨平台開發、資料交換與軟體維護。

---

### 2. 如果每個作業系統都使用不同的時間表示法，撰寫跨平台程式會遇到哪些困難？

假設：

```text
Windows：以 1601-01-01 為起點
Linux   ：以 1970-01-01 為起點
macOS   ：以另一日期為起點
```

則同一個數值：

```text
1000000000
```

在不同系統上可能代表不同時間。

例如：

```python
save_time = get_time()
```

在 Windows 與 Linux 之間交換資料時，可能會發生：

```text
時間解讀錯誤
```

甚至導致：

- 檔案時間錯誤
- 日誌（log）排序錯誤
- 資料庫紀錄錯誤
- 網路服務同步失敗

因此程式設計師必須撰寫大量轉換程式碼：

```text
Windows Time
    ↓
Unix Time
    ↓
Linux Time
```

增加系統複雜度與錯誤風險。

---

### 3. Python 的 `time` 模組在不同作業系統之間扮演了什麼角色？

`time` 模組提供了一個統一的介面（Unified Interface）。

程式設計師只需要：

```python
import time

print(time.time())
```

而不必知道底層使用的是：

```text
Windows API
Linux System Call
macOS System Call
```

概念上：

```text
Python 程式
    ↓
time 模組
    ↓
作業系統 API
    ↓
Windows / Linux / macOS
```

因此：

> `time` 模組隱藏了不同作業系統的實作細節，讓程式設計師使用相同的方式取得時間資訊。

這正是程式語言函式庫的重要功能之一。

---

### 4. 這是否可以視為抽象化（Abstraction）的一個例子？請說明理由。

是的，這是一個典型的抽象化（Abstraction）例子。

所謂抽象化，是指：

> 隱藏底層細節，提供一致且容易使用的介面。

例如：

```python
time.time()
```

看起來只是一個簡單的函式呼叫。

但實際上可能經過：

```text
Python 程式
    ↓
time 模組
    ↓
作業系統 API
    ↓
系統時間服務
    ↓
RTC（Real-Time Clock）
    ↓
硬體時鐘
```

程式設計師不需要了解：

- RTC 如何運作
- Windows 如何儲存時間
- Linux 如何取得系統時間
- 時鐘晶片如何計數

只需要：

```python
time.time()
```

即可取得時間資訊。

因此：

> `time` 模組將複雜的作業系統與硬體細節隱藏起來，提供簡單一致的介面，這正是抽象化的典型例子。
