函式（Function）的基本語法如下：

```python
def function_name(parameters):
    # Function body
```
其中：
- def：定義函式（define function）。
- function_name：函式名稱。
- parameters：參數（可省略）。
- :：表示函式主體（function body）的開始。
- 縮排（indentation）部分為函式主體。


若函式需要回傳結果，則可使用 return：
```python
def function_name(parameters):
    # Function body
    return expression
```
其中：
- return 用來結束函式的執行。
- expression 為運算式（expression）。
- Python 會先計算（evaluate）expression，再將其值（value）回傳給呼叫者。

#### 呼叫堆疊

每當函式被呼叫，系統便會建立啟動記錄(activation record), 又被稱作啟動框架(activation frame)，用來儲存函式參數與變數，並將啟動記錄放置於被稱做呼叫堆疊（calls tack）的記憶體區段內。

呼叫堆疊又被稱作執行堆疊（executi on stack）、執行期間堆疊(runtime stack)，或機器堆疊(machine stack) ，通常簡稱為堆疊（stack）。（註：資料結構）當某函式呼叫另—個函式，呼叫者的啟動記錄維持不動，接著為被呼叫的新函式建立新的啟動記錄。
當該函式完成其工作，將控制權還給呼叫者，其啟動記錄便會從呼叫堆疊裡移除。

呼叫堆疊以後進先出法(last-in, first-out)儲存啟動記錄：最後被呼叫函式的啟動記錄會先從堆疊裡移除。
舉個例子，假設函式ml 呼叫函式m2，接著呼叫m3 。
執行期間系統會將ml 的啟動記錄放進堆疊裡，接著再將m2 的啟動記錄放進堆疊，最後才是m3 。
函式m3 被執行後，其啟動記錄會先從堆疊裡被移除。m2 完成其工作後，其啟動記錄接著從堆疊裡被移除。
最後，ml 執行後，其啟動記錄便會從堆疊裡被移除。

了解呼叫堆疊能幫助您理解函式被呼叫的方式。

#### 參數的基本機制

使用函式時，可透過參數（Parameter）與引數（Argument）進行資料的傳遞與接收。在介紹各種參數型式之前，先了解下列兩個名詞：

- **實際引數（Actual Argument）**：呼叫函式時所提供的資料或物件，會傳遞給函式的形式參數。例如：

  ```python
  printGrade(95)
  ```

  其中 `95` 為實際引數。

- **形式參數（Formal Parameter）**：定義函式時所宣告的參數，用來接收實際引數所傳遞的資料。例如：

  ```python
  def printGrade(score):
      ...
  ```

  其中 `score` 為形式參數。

除了最常見的位置引數（Positional Arguments）之外，Python 還提供下列參數機制：

- **預設參數值（Default Parameter Values）**：定義函式時可為形式參數指定預設值。若呼叫函式時未提供對應的實際引數，則使用預設值。

  ```python
  def greet(name="Guest"):
      print("Hello,", name)
  ```

- **關鍵字引數（Keyword Arguments）**：呼叫函式時，可直接使用參數名稱指定對應的值，提高程式的可讀性。

  ```python
  greet(name="Tom")
  ```

- **可變長度參數（Variable-Length Arguments）**：

  - `*args`：收集額外的位置引數，並以 Tuple 的形式儲存。
  - `**kwargs`：收集額外的關鍵字引數，並以 Dictionary 的形式儲存。

  ```python
  def demo(*args, **kwargs):
      print(args)
      print(kwargs)
  ```

- **解包運算子（Unpacking Operators）**：

  - `*` 可將可迭代物件（Iterable）中的元素解包成多個位置引數。
  - `**` 可將映射物件（Mapping，例如 Dictionary）中的鍵值對解包成多個關鍵字引數。

  ```python
  numbers = [1, 2, 3]
  print(*numbers)
  ```

  ```python
  info = {"name": "Tom", "age": 18}
  print(**info)
  ```

透過上述機制，函式可以更靈活地接收與傳遞資料。

### Python 函式參數的種類

```python

    positional_only_parameters,

):
    ...

| 語法 | 種類 | 說明 |
|--------|--------|--------|

| `/` | Positional-Only Separator | `/` 前面的參數只能使用位置引數傳遞。 |
| `positional_or_keyword_parameters` | Positional-or-Keyword Parameters | 可使用位置引數或關鍵字引數（Keyword Arguments）傳遞。 |
| `*var_positional_parameter` | Variable Positional Parameter | 使用 `*` 收集額外的位置引數，並以 Tuple 儲存。 |
| `keyword_only_parameters` | Keyword-Only Parameters | 必須以關鍵字引數傳遞的參數。 |
| `**var_keyword_parameter` | Variable Keyword Parameter | 使用 `**` 收集額外的關鍵字引數，並以 Dictionary 儲存。 |

```

# 形式參數的 `*` 星號運算式

在 Python 中，`*` 運算子最常見的用途是乘法運算：

```python
a = 3 * 5
print(a)
```

輸出：

```text
15
```

不過在函式的參數列（parameter list）中，`*` 具有不同的用途。它可以收集額外的位置引數（positional arguments），使函式能夠接受不固定數量的引數。

其基本語法如下：

```python
def function_name(parameter1, parameter2, ..., parameterN, *args):
    # 函式主體
```

例如：

```python
def show(a, b, *tp):
    print(f'a = {a}')
    print(f'b = {b}')
    print(f'tp = {tp}')
```

呼叫：

```python
show(10, 20, 30, 40, 50)
```

輸出：

```text
a = 10
b = 20
tp = (30, 40, 50)
```

在此範例中：

- `a` 接收第一個位置引數。
- `b` 接收第二個位置引數。
- `*tp` 收集其餘的位置引數。

因此：

```python
tp == (30, 40, 50)
```

可以看到：

- `*tp` 用來收集額外的位置引數。
- `*tp` 會將多餘的位置引數收集成一個 tuple。
- `tp` 只是變數名稱，可以改成任何合法識別字，例如：

```python
def show(*numbers):
    print(numbers)
```

或：

```python
def show(*values):
    print(values)
```

效果完全相同。

````python
# 自訂 Mapping 類別 後續會談到
from collections.abc import Mapping

class MyMapping(Mapping):

    def __init__(self):
        self.data = {
            "name": "Tom",
            "age": 18
        }

    def __getitem__(self, key):
        return self.data[key]

    def __iter__(self):
        return iter(self.data)

    def __len__(self):
        return len(self.data)

def student(name, age):
    print(name, age)

info = MyMapping()

student(**info)
````



## 引數如何傳遞？

在討論形式參數（formal parameter）之前，先來了解呼叫函式時，實際引數（actual argument）是如何傳遞給函式的。

簡單來說，就是：

- 呼叫函式時提供引數。
- 函式接收這些引數並執行相應的程式碼。

大多數程式語言常以兩種方式來描述引數傳遞：

- **傳值（Call by Value）**：
  傳遞的是資料的副本，因此在函式內修改參數，不會影響原來的資料。

- **傳址（Pass-by-Reference）**：
  傳遞的是資料的參照（或記憶體位址），因此在函式內修改參數，可能會影響原來的資料。

那麼 Python 採用哪一種方式呢？

事實上，Python 的引數傳遞機制與上述兩種說法都不完全相同。Python 採用的是 **Call by Object Reference**（物件參照呼叫），也常稱為 **Call by Assignment**。

當函式被呼叫時，實際引數所參照的物件會指定給對應的形式參數。因此，函式內的參數名稱與呼叫端的變數名稱可能同時參照同一個物件。

例如：

```python
def func(x):
    print(id(x))

n = 10
print(id(n))
func(n)


輸出結果中的兩個位址相同，表示參數 x 與變數 n 參照同一個物件。

然而Python 中物件是否可變（mutable）會影響函式呼叫後的結果。

- 不可變（Immutable）物件

常見的不可變物件包括：int、float、bool、str、tuple。由於不可變物件建立後無法修改，因此在函式內重新指定參數時，不會影響呼叫端的變數。

def add_one(x):
    x = x + 1
    print("函式內 x =", x)

n = 10
add_one(n)

print("函式外 n =", n)

輸出：

函式內 x = 11

函式外 n = 10


雖然 x 與 n 起初參照同一個整數物件，但 x = x + 1 會建立新的整數物件，因此不會改變原來的 n。

- 可變（Mutable）物件

常見的可變物件包括：list、dict、set。如果函式直接修改物件內容，呼叫端也會看到修改結果。

def append_item(data):
    data.append(100)

numbers = [1, 2, 3]

append_item(numbers)

print(numbers)


輸出：

[1, 2, 3, 100]


這是因為函式中的參數 data 與變數 numbers 參照同一個串列物件，而 append() 是直接修改該物件的內容。

````python
# 課堂練習
def triple(x, y):                      
    x = x * 3                          
    y = y * 3                          
    print('執行 triple() 函式 ------') 
    print(f'x = {x}     y = {y}\n')    
                                       
                                       
x = 10                                 
A = [2, 4, 6, 8]                       
print('呼叫 triple() 函式前 ------')   
print(f'x = {x}     A[1] = {A[1]}\n')  
triple(x, A[1])    # 呼叫 triple() 函式
print('呼叫 triple() 函式後 ------')   
print(f'x = {x}     A[1] = {A[1]}')    
````

**Python 是 Call by Object Reference**

#### 為什麼 `x` 和 `A[1]` 都沒有改變？
 在呼叫

 ```python
 triple(x, A[1])
 ```

 時，形式參數 `x` 與 `y` 分別參照與實際引數相同的物件。

 由於 `int` 屬於不可變（immutable）物件，敘述

 ```python
 x = x * 3
 y = y * 3
 ```
 不會修改原來的整數物件，而是使 `x` 與 `y` 改為參照新的整數物件。因此函式結束後，呼叫端的 `x` 與 `A[1]` 不受影響。

### 可變物件的對照範例

```python
def triple(data):
    data[1] *= 3

A = [2, 4, 6, 8]

print(A)

triple(A)

print(A)
```

輸出：

```text
[2, 4, 6, 8]
[2, 12, 6, 8]
```

此時函式直接修改串列物件的內容，因此呼叫端的 `A` 也跟著改變。

這是因為函式中的參數 `data` 與呼叫端的變數 `A` 參照同一個串列物件，而串列（`list`）屬於可變物件（mutable object），其內容可以在原地（in-place）修改。

````python
def triple(lst):                      
    for i in range(len(lst)):         
        lst[i] = lst[i] * 3           
    print('執行 triple() 函式 ------')
    print(f'串列 lst = {lst}\n')      
                                      
                                      
arr = [2, 4, 6, 8, 10]                
print('呼叫 triple() 函式前 ------')  
print(f'串列 arr = {arr}\n')          
triple(arr)    # 呼叫 triple() 函式   
print('呼叫 triple() 函式後 ------')  
print(f'串列 arr = {arr}')            
````

### 為什麼 `arr` 被改變了？

呼叫：

```python
triple(arr)
```

時，形式參數 `lst` 與實際引數 `arr` 參照同一個串列物件。

函式開始執行時，可以想成：

```python
lst ──┐
      ▼
[2, 4, 6, 8, 10]
      ▲
      │
arr ──┘
```

因此：

```python
lst[i] = lst[i] * 3
```

並不是建立新的串列，而是在原來的串列物件中直接修改元素內容。

例如第一次迴圈：

```python
lst[0] = lst[0] * 3
```

會將串列改成：

```python
[6, 4, 6, 8, 10]
```

由於 `arr` 和 `lst` 參照的是同一個串列物件，因此透過 `lst` 所做的修改，`arr` 也會看到相同的結果。

---

### 物件參照示意圖

呼叫函式後：

```text

      ▼
[2, 4, 6, 8, 10]
      ▲
      │
lst ──┘
```

執行完成後：

```text

      ▼
[6, 12, 18, 24, 30]
      ▲
      │
lst ──┘
```

可以看到改變的是同一個串列物件，因此函式結束後，呼叫端的 `arr` 也被改變了。

### 與整數範例的差異

前面的範例：

```python
def triple(x, y):
    x = x * 3
    y = y * 3
```

中的 `x`、`y` 是整數（`int`），而 `int` 屬於不可變物件（immutable object）。

因此：

```python

```

不會修改原來的物件，而是讓 `x` 改為參照新的整數物件。

然而本例中的串列（`list`）是可變物件（mutable object），因此：

```python
lst[i] = lst[i] * 3
```

是在原來的串列物件上直接修改內容，所以呼叫端的 `arr` 也會跟著改變。

- Python 採用 **Call by Object Reference（Call by Assignment）**。
- 呼叫函式時，不會自動複製整個串列。
- `lst` 與 `arr` 參照同一個串列物件。
- `list` 為可變物件（mutable object）。
- 修改串列元素會直接改變原物件內容。
- 因此函式執行後，呼叫端的 `arr` 也會被改變。

#### Python 的命名空間（Namespace）與作用域（Scope）

無論是變數、函式、類別，甚至模組，在 Python 中都會以名稱（Name）來存取對應的物件（Object）。

- 命名空間（Namespace）：名稱（Name）與物件（Object）之間的對應關係，是「名稱與記憶體中物件的對應表」。
- 作用域（Scope）：某個名稱可以被存取或使用的程式區域。

當程式引用一個名稱時，Python 會依照 LEGB 規則（LEGB Rule）在不同的命名空間中搜尋該名稱：

Local → Enclosing → Global → Built-in
1. Local（區域作用域）
- 函式內部定義的名稱。
- 只在目前函式執行期間可見。
- 對應的命名空間稱為 Local Namespace。

2. Enclosing（包圍作用域）
- 巢狀函式（nested functions）中，外層函式的區域作用域。
- 供內層函式存取外層函式的名稱。
- 對應的命名空間稱為 Enclosing Namespace。

3. Global（全域作用域）
- 在模組（.py 檔案）層級定義的名稱。
- 可在同一模組中的各個函式使用。
- 對應的命名空間稱為 Global Namespace。

4. Built-in（內建作用域）
- Python 內建名稱所在的作用域。
- 由 builtins 模組提供，例如 print()、len()、range() 等函式。
- 對應的命名空間稱為 Built-in Namespace。

名稱查找順序

當 Python 遇到一個名稱時，會依序在下列命名空間中搜尋：

1. Local Namespace
2. Enclosing Namespace
3. Global Namespace
4. Built-in Namespace

找到後便停止搜尋；若所有命名空間都找不到，則拋出 NameError。


- Namespace（命名空間）：名稱儲存的位置。
- Scope（作用域）：名稱可見的範圍。
- Memory（記憶體）則是實際存放物件的地方。

例如：
```python
x = 10      # x 位於 Global Namespace
def f():
    y = 20  # y 位於 Local Namespace
    print(x)
f()
```
其中：
- x 存在於 Global Namespace，其作用域為整個模組。

## Python 的垃圾回收（Garbage Collection, GC）

當程式執行時，Python 會在記憶體中建立各種物件（Object）。

```python
x = 10
```

例如上面的程式碼會建立整數物件 `10`，並讓名稱 `x` 參考（reference）該物件。

```text
Global Namespace

x ───► 10
```

當一個物件不再被任何名稱或其他物件參考時，Python 便可以回收該物件所佔用的記憶體，此過程稱為垃圾回收（Garbage Collection, GC）。

---

## 參考計數（Reference Counting）

Python 主要透過參考計數來管理記憶體。

每個物件都會維護一個參考計數器（reference count），記錄目前有多少個參考指向該物件。

例如：

```python
x = [1, 2, 3]
```

```text
x ───► [1, 2, 3]

Reference Count = 1
```

接著：

```python
y = x
```

```text
x ──┐
    │
    ▼
 [1,2,3]
    ▲
    │
y ──┘

Reference Count = 2
```

此時 `x` 與 `y` 都參考同一個串列物件。

---

## 參考計數歸零

若移除其中一個參考：

```python
del x
```

```text
y ───► [1,2,3]

Reference Count = 1
```

物件仍然存在。

若再移除最後一個參考：

```python
del y
```

```text
Reference Count = 0
```

此時 Python 便可釋放該物件所佔用的記憶體。

---

## Local Namespace 與物件回收

考慮下面的程式：

```python
def f():
    x = [1, 2, 3]

f()
```

函式執行時：

```text
Local Namespace

x ───► [1,2,3]
```

函式結束後：

```text
Local Namespace 被銷毀
↓
名稱 x 消失
↓
物件失去最後一個參考
↓
Reference Count = 0
↓
物件被回收
```

因此，名稱（Name）與物件（Object）是不同的概念。

---

## 為什麼還需要 Garbage Collector？

參考計數有一個問題：

### 循環參考（Circular Reference）

```python
a = []
b = []

a.append(b)
b.append(a)
```

```text
a ───► []
 ▲      │
 │      ▼
 └──── b
```

即使：

```python
del a
del b
```

兩個串列仍然互相參考：

```text
Object A ─► Object B
     ▲         │
     └─────────┘
```

因此參考計數不會降到 0。

只靠 Reference Counting 無法回收這類物件。

---

## Garbage Collector（GC）

為了解決循環參考問題，Python 額外提供 Garbage Collector（GC）。

GC 會定期檢查物件之間的參考關係：

```text
Object A ↔ Object B
```

若發現：

- 外部已無法存取這些物件。
- 只剩下彼此互相參考。

則 GC 會將它們回收。

---

## 檢視 GC 模組

Python 提供內建模組：

```python
import gc
```

查看 GC 是否啟用：

```python
import gc

print(gc.isenabled())
```

手動執行一次垃圾回收：

```python
gc.collect()
```

---

## Namespace、Scope 與 GC 的關係

```text
Name
  ↓
Namespace
  ↓
Object
  ↓
Memory
```

- **Namespace（命名空間）**：名稱與物件的對應表。
- **Scope（作用域）**：名稱可見的範圍。
- **Object（物件）**：實際存在於記憶體中的資料。
- **GC（Garbage Collector）**：負責回收不再需要的物件。

當函式結束時：

```text
Local Scope 消失
↓
Local Namespace 被銷毀
↓
名稱消失
↓
物件失去參考
↓
Reference Count = 0
↓
物件被回收（或等待 GC 回收）
```

---

## 重點整理

- **Scope 決定名稱看不看得到。**
- **Namespace 決定名稱對應到哪個物件。**
- **Object 實際存放在記憶體中。**
- **GC 負責回收不再被參考的物件。**

當程式使用某個名稱時，Python 會依照 **LEGB 規則**：

```text
Local → Enclosing → Global → Built-in
```

在不同的命名空間中搜尋該名稱；找到後，再透過該名稱取得記憶體中的物件。
