將提交上傳到 GitHub Repository：`git push`

#### Python 與 Python Interpreter

當我們安裝並執行 Python 時，是透過某個 Python Interpreter 來執行 Python 程式。平常我們說Python，其實這個詞有兩種不同的意思。第一個意思是Python 語言（Python Language），它負責定義語法（Syntax）、關鍵字（Keywords）、資料型別（Data Types）、函式的行為、語言規則（Language Rules）。例如
```python
print("Hello World!")
```
是一段合法的 Python 程式。然而
```python
print("Hello World!"
```
少了一個右括號，因此不符合 Python 語法。

這些規則是由 Python 語言本身所定義的。

第二個意思是Python Interpreter，它是一個真正的程式（Program），負責讀取並執行 Python 程式碼。

有一門課叫編譯器（Compiler），主要研究如何讓電腦理解程式語言？換句話說
```text
Python Language
        ↓
如何檢查語法？

如何理解程式的意思？

如何執行程式？

如何產生機器碼？
```
這些問題就是編譯器課程的核心。

舉例來說，`print("Hello World!")`。編譯器（Compiler）或直譯器（Interpreter）並不會一開始就理解這段程式的意義。相反地，它會逐步分析程式碼。第一層字元（Characters）。首先，程式碼只是一串字元：`p r i n t ( " H e l l o ...`，這些字元是什麼？這些字元本身並沒有太多意義，編譯器必須先將它們組合成更有意義的單位。

第二層詞元（Tokens）。接著進行詞法分析（Lexical Analysis），將字元組合成詞元（Token）：
```text
print
(
"Hello World!"
)
```
其中`print`是一個識別字（Identifier）、(和)是括號符號、"Hello World!"是一個字串常值（String Literal）。

詞法分析的工作就是將字元串轉換成一連串有意義的詞元（Tokens）。

第三層語法結構（Syntax）。接著進行語法分析（Syntax Analysis）。編譯器會檢查`print("Hello World!")`，是否符合 Python 文法。例如`print("Hello World!"`，缺少右括號。因此產生Syntax Error，因為這段程式不符合 Python 的語法規則。

第四層語意，理解`print(...)`的意思是，呼叫 print 函式。這稱為語意分析Semantic Analysis。

第五層：執行。最後顯示 Hello World!

CPython 是Implementation（實作）。編譯器課通常會進一步研究：
```text
Python Language
        ↓
CPython 如何實作？

Java Language
        ↓
JVM 如何實作？

C Language
        ↓
GCC 如何實作？
```

總之，編譯器課程主要研究：「如何讓電腦理解並執行程式語言？」因此其重點通常在於程式語言的規格（Specification）及其實作（Implementation）之間的轉換過程，例如詞法分析、語法分析、語意分析與程式執行。其中：
- Python Language 屬於規格（Specification）。
- CPython、PyPy 等屬於規格的實作（Implementation）。

#### 常見的 Python Interpreter


##### CPython

最常見的 Python Interpreter。
- Python 官方實作（Reference Implementation）
- 使用 C 語言撰寫
- 幾乎所有 Python 教材預設都是 CPython
- GitHub Codespaces 通常也是使用 CPython

```text
Python 程式
      ↓
 CPython
      ↓
 作業系統
```
##### PyPy

另一種 Python 實作。

- 使用 RPython 撰寫
- 具有 JIT（Just-In-Time Compilation）功能
- 某些程式執行速度可能比 CPython 更快

```text
Python 程式
      ↓
   PyPy
      ↓
 JIT 編譯
      ↓
 作業系統
```
##### Jython

Java 平台上的 Python。

- 使用 Java 撰寫
- 執行於 JVM（Java Virtual Machine）
- 可直接使用 Java 類別庫

```text
Python 程式
      ↓
   Jython
      ↓
    JVM
      ↓
 Java 平台
```
##### IronPython

.NET 平台上的 Python。

- 使用 C# 實作
- 執行於 .NET CLR
- 可使用 .NET 類別庫

```text
Python 程式
      ↓
 IronPython
      ↓
    CLR
      ↓
 .NET 平台
```
##### MicroPython

嵌入式系統版本的 Python。

例如：

- ESP32
- Raspberry Pi Pico
- 微控制器

```text
Python 程式
      ↓
 MicroPython
      ↓
 微控制器
```
常見於 IoT 與嵌入式開發

#### 記憶體（Memory）
在程式執行期間，資料通常會暫時儲存在記憶體（Memory）中。你可以把記憶體想像成電腦工作時使用的草稿紙或工作桌面。當程式需要資料時，
可以從記憶體取出來使用。例如：
```text
記憶體
┌─────────────┐
│Hello World! │
└─────────────┘
```

#### 變數（Variable）

如果記憶體中有許多資料，Python 要如何找到其中某一筆資料呢？因此我們可以替資料取一個名稱：
```python
message = "Hello World!"
```
可以想像成：
```text
記憶體
┌─────────────┐
│Hello World! │
└─────────────┘
       ▲
       │
   message
```
其中：

- `"Hello World!"` 是資料。
- `message` 是資料的名稱。

未來如果需要這筆資料，就不必重新輸入`"Hello World!"`，而可以直接使用`message`來存取它。接著`print(message)`，Python 會根據名稱 `message` 找到對應的資料，並顯示`Hello World!`。

#### 記憶體與儲存裝置

事實上，電腦中有不同的儲存空間。例如：`主記憶體（RAM, Random Access Memory）以及固態硬碟（SSD）或硬碟（Hard Disk）。當我們撰寫：

```python
message = "Hello World!"
```
時，這段程式碼最初是儲存在檔案中：`hello.py`或`notebook.ipynb`，而這些檔案通常存放在`SSD 或硬碟`上。當按下執行按鈕後：
```text
程式檔
    ↓
Python Interpreter
    ↓
記憶體（RAM）
    ↓
CPU 執行
```
程式需要使用的資料才會被載入記憶體。因此：`message = "Hello World!"`在程式執行期間，通常可以把它想像成位於記憶體中。

#### 為什麼需要記憶體？

因為 CPU 執行程式時，必須能快速取得資料。可以把它想像成：
```text
書櫃（SSD）
    ↓
存放大量資料

書桌（RAM）
    ↓
目前正在使用的資料
```
通常會先把要閱讀的書從書櫃拿到書桌上，再開始閱讀。電腦也有類似的運作方式：
```text
SSD
    ↓
載入
    ↓
RAM
    ↓
CPU 執行
```
未來學習作業系統與計算機組織時，會深入探討：

- 記憶體（Memory）
- 快取（Cache）
- SSD
- 虛擬記憶體（Virtual Memory）

#### 陳述式與運算式
剛剛我們已經寫過：
```python
message = "Hello World!"
print(message)
```
這兩行程式碼可以幫助我們認識 Python 中兩個重要的概念：
```text
陳述式（Statement）

運算式（Expression）
```
在 Python 中，每一行完成一個動作的程式碼，通常稱為一個陳述式（Statement）。例如：
```python
message = "Hello World!"
```
以及：
```python
print(message)
```
都是陳述式。

另一方面，如果一段程式碼可以產生一個值，那麼這段程式碼就稱為運算式（Expression）。例如：
```python
"Hello World!"
```
是一個運算式。它所代表的值為`Hello World!`。再例如`2 + 3`是一個運算式。其值為5。

回到剛剛的程式：`message = "Hello World!"`，觀察`"Hello World!"`是運算式（Expression）。它產生的值為Hello World!，而整行`message = "Hello World!"`則是一個陳述式（Statement），其作用是將運算式所得到的值存放到變數 `message` 中。


接著看`print(message)`，其中`message`也是一個運算式。其值為`Hello World!`。而整行`print(message)`則是一個陳述式，其作用是將變數中的內容顯示在畫面上。
#### 圖解

```text
message = "Hello World!"

          │
          │
          ▼

      運算式
   "Hello World!"

          │
          ▼

      Hello World!

          │
          ▼

      陳述式
message = ...
```

#### 函式（Function）的概念
看到下面的程式：`print("Hello World!")`，是否覺得看起來有點像數學中的$f(x)$，其中：
- `f` 類比為 `print`
- `x` 類比為 `"Hello World!"`
在數學中，$f(x)$表示把輸入的資料 \(x\) 交給函式 \(f\) 處理。例如$f(x)=x^2$，則$f(3)=9$，可以表示成：
```text
輸入
 3
 │
 ▼
 f
 │
 ▼
 9
```
Python 中也有類似的概念`print("Hello World!")`，其中：
- `print` 是函式名稱（Function Name）。
- `"Hello World!"` 是提供給函式的資料。
- 括號 `( )` 用來放入要交給函式處理的資料。
可以表示成：
```text
輸入
"Hello World!"
       │
       ▼
     print
       │
       ▼
Hello World!
```
函式可以想像成一台機器：
```text
資料
  │
  ▼
函式
  │
  ▼
結果
```
例如：`print("Python")`輸出：Python；`print(100)`，輸出：100。因此`print("Hello World!")`可以理解成：將資料 `"Hello World!"` 交給函式 `print` 處理，而 `print` 的工作就是把資料顯示在畫面上。

#### Python 的做法
在 Python 中，變數就像資料的名稱。
```text
message
    ↓
"Python"
```
或
```text
message
    ↓
100
```
因此：變數本身不需要先宣告資料型別。Python 會根據資料內容，自動判斷資料型別。這種方式稱為動態型別（Dynamic Typing）

#### 與 C 語言比較

在 C 語言中，通常必須先宣告資料型別：

```c
int message = 100;
```
表示：
```text
message 是整數變數
```
若要存放文字：
```c
char message[] = "Python";
```
則需要重新使用另一種型別。

因此在 C 語言中：
```text
變數
    │
    ▼
型別固定
```
這種方式稱為靜態型別（Static Typing）。
