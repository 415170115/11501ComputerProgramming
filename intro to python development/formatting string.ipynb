### you can combine string with +

```python
first_name = 'Chirstopher'
last_name = 'Harrison'
print(first_name + last_name)
print('Hello '+ first_name + ' ' + last_name)
```

### custom string formatting

```python
output ='Hello, ' + first_name + ' ' + last_name
output ='Hello, {} {}'.format(first_name, last_name)
output = 'Hwllo, {0} {1}'.format(first_name, last_name)
# only available in python 3
output = f'Hello, {first_name} {last_name}'
```

```python
first_name = 'Chirstopher'
last_name = 'Harrison'

output ='Hello, ' + first_name + ' ' + last_name
print(output)
output ='Hello, {} {}'.format(first_name, last_name)    #{}{}為預留參數位置，分別對應到first_name and last_name
print(output)
output = 'Hwllo, {0} {1}'.format(first_name, last_name)     #若想特別指定位置，可在大括號中放數字 0,1，對應到第一個與第二個變數，btw計數從0開始
print(output)
output = 'Hwllo, {1}, {0}'.format(first_name, last_name)
print(output)
# only available in python 3
output = f'Hello, {first_name} {last_name}'     # f means format，直接把變數放在大括號裡
print(output)
```
