### string can be stored in variable

```python
first_name = 'Chistopher'
print(first_name)
```

### you can combine strings with +

```python
first_name = 'Chistopher'
last_name = 'Harrison'
print(first_name + last_name)
print('Hello ' + first_name + ' '+ last_name)
```

### you can use functions to modify strings

```python
sentence = 'The dog is named Sammy'
print(sentence.upper())
print(sentence.lower())
print(sentence.capitalize())
print(sentence.count('a'))      #計算'a'在句子裡出現幾次
```

### the functions helps us format strings we save to files and database, or display to user

```python
first_name = input('What is your first name?')
last_name = input('What is your last name?')
print('Hello ' + first_name.capitalize() + ' '\
      + last_name.capitalize())
```
