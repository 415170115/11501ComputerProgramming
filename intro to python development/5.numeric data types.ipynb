### numbers can be stored in variables

```python
pi = 3.14159
print(pi)
```

### you can do math with numbers

```python
first_num = 6
second_num = 2
print(first_num + second_num)
print(first_num ** second_num)

# + addition; - subtraction; * multiplication; / division; ** exponent
```

### if you combine strings with numbers, Python gets confused

```python
days_in_feb =28
print(days_in_feb + ' days in February')
# number + number; string + string 
```

### when displaying a string that contains numbers you must convert the numbers into strings

```python
days_in_feb =28
print(str(days_in_feb )+ ' days in February')
```

### numbers can be stored as striings; numbers stored as string are treated as string

```python
first_num = '5'
second_num ='6'
print(first_num + second_num)
```

### the input function always returns strings

```python
first_num = input('enter first number')
second_num = input('enter second number')
print(first_num + second_num)
```

```python
first_num = input('enter first number')
print('enter first_number' + ' '+first_num)
second_num = input('enter second number')
print('enter second_number'+ ' '+second_num)
print(first_num + second_num)
```

### numbers stored as strings must be ocnverted to numeric values before doing math

```python
first_num = input('enter first number')
second_num = input('enter second number')
print(int(first_num)+ int(second_num))
print(float(first_num)+ float(second_num))
```

```python
first_num = input('enter first number')
second_num = input('enter second number')
print('enter first_number' + ' '+first_num)
print('enter second_number'+' '+ second_num)
print(int(first_num)+ int(second_num))
print(float(first_num)+ float(second_num))
```
