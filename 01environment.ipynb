為自己學Python

安裝 Python 的困難度不高，基本上就是到 Python 的官方網站下載安裝程式並照著指示做就行了，目前主流的作業系統，例如 Windows、macOS 以及 Linux 都有相對應的版本。是說，如果各位使用的是 macOS 跟 Linux 作業系統，可能會發現原本電腦裡就已經直接內建 Python 了，這樣還需要另外安裝嗎？作業系統內建安裝的 Python 通常會比現行版本稍微舊一點，如果單純就以體驗或寫一些簡單的程式的話是夠用的。作業系統之所以會內建安裝 Python 是有原因的，因為有些系統程式本身就是用 Python 開發的。Python 的版本從早期的 2.x 版到現在的 3.x 版，除了正宗的 CPython 之外，還有其他分支的實作品版本（像是 Jython、IronPython、PyPy 等），如果不想把自己的電腦環境弄的亂七八糟，或是因為專案需要常常安裝不同版本的 Python，通常會建議另外使用版本管理工具來建構隔離的開發環境，這裡將使用 pyenv 做為範例。

可以把 pyenv 想像成一個工具箱，在這個工具箱裡會可以放很多種工具，但這個工具箱裡如果沒有任何工具它就只是個空的盒子而已。pyenv 不只是個工具箱，還是有分隔板的工具箱，不同版本的 Python 會被安裝在不同的隔板裡，安裝套件的時候也會安裝在各自的隔板裡不會互相干擾，透過 pyenv 就能做到簡單的切換 Python 的環境及版本。

提醒一下，如果你的作業系統（例如 macOS）原本就有預先安裝 Python 的話，即使透過 pyenv 安裝了其他版本的 Python，原本系統內建的 Python 也建議不要移除，它存在作業系統裡一定有原因的，你不會知道系統裡有哪些程式正在使用它。

pyenv 的主要功能包括：

- 安裝多個 Python 版本
- 切換全域（global）Python 版本
- 為不同專案指定 Python 版本
- 管理 CPython、PyPy 等不同 Python 發行版本

#### 安裝 pyenv
安裝 pyenv 的方式滿簡單的，只要到 GitHub 的說明頁面，照著指示複製貼上應該一下子就能安裝完成。如果是 Windows 作業系統則是另外安裝 pyenv-win：

1. pyenv 的主要用途是什麼？

      pyenv 是一個 Python 版本管理工具，可以在同一台電腦上安裝與切換多個 Python 版本，例如 Python 3.10、3.11、3.12、3.13，方便不同專案使用不同版本，而不會互相衝突。

2. 為什麼有些開發者會在同一台電腦安裝多個 Python 版本？

      因為不同專案可能要求不同的 Python 版本。

       例如：

         專案 A 只能使用 Python 3.10
         專案 B 使用 Python 3.13 的新功能
         舊系統可能仍依賴 Python 3.9

      如果只安裝一個版本，可能導致部分程式無法執行，因此需要透過 pyenv 管理多個版本。

3. 請說明下列指令的用途：

   (a) `pyenv versions`

      顯示目前已安裝的 Python 版本，以及目前正在使用的版本。

      例如：
         $ pyenv versions
         system
         3.10.15
         * 3.12.7
      * 表示目前使用中的版本。

   (b) `pyenv install 3.12.7`

      下載並安裝 Python 3.12.7。

      例如：pyenv install 3.12.7

      安裝後即可切換到此版本。

   (c) `pyenv shell 3.12.7`

      暫時將目前終端機的 Python 版本切換成 3.12.7。

      pyenv shell 3.12.7

      關閉終端機後設定失效。

   (d) `pyenv global 3.12.7`

      將整台電腦預設 Python 版本設定為 3.12.7。

      pyenv global 3.12.7

      之後開啟任何新終端機都會優先使用此版本。

   (e) `pyenv local 3.12.7`

      只在目前專案資料夾使用 Python 3.12.7。

      pyenv local 3.12.7

      會建立一個：.python-version檔案。

      當進入該資料夾時，自動使用指定版本。

4. `shell`、`local` 與 `global` 三種設定方式有何差異？

   指令 ｜作用範圍｜持續時間

   shell|目前終端機｜終端機關閉後失敗
   local|目前專案資料夾｜長期有效
   global|整台電腦預設｜長期有效

   優先順序：

      shell
      ↓
      local
      ↓
      global
      ↓
      system

   越上面優先權越高。

#### 開發環境

開發環境（Development Environment）是影響程式開發效率的重要因素之一。

過去，開發者通常需要自行安裝 Python、設定編輯器、管理套件以及維護開發環境。然而，現今已可透過雲端開發環境（Cloud Development Environment）快速建立一致且可重現的開發平台。

本課程將使用 GitHub Codespaces 作為主要開發環境。學生只需具備 GitHub 帳號，即可透過瀏覽器使用 Visual Studio Code、Python 與 Linux Terminal 進行程式開發。

現代 Python 開發環境通常包含：

- Python Interpreter（Python 直譯器）
- pip Package Manager（套件管理器）
- Virtual Environment（虛擬環境，venv）
- Visual Studio Code（VS Code）
- Git 與 GitHub（版本控制）
- AI Assistant（ChatGPT、Claude、Gemini、GitHub Copilot）
- VS Code Extensions（Python、Pylance、Jupyter 等延伸模組）
- Container Technologies（Docker 等容器技術）

其中，VS Code 的延伸模組Pylance能協助檢查程式中的潛在問題，例如：

- 未定義的變數
- 函式參數數量不正確
- 常見語法錯誤
- 型別相關問題

這類工具稱為靜態分析工具（Static Analysis Tools），可在程式執行前協助發現錯誤。

當 VS Code 顯示錯誤訊息或警告時，可以進一步利用 ChatGPT 或 GitHub Copilot 協助理解錯誤原因、分析問題並提出修正建議。

#### 套件和虛擬環境

在大多數其他的程式語言之中， 一個套件( package ) 相當於一個程式庫(library)，都是由一組程式碼組成。Python 因為「一切需要的都已包含在內」而聞名， 大部分功能只需透過簡單的import陳述式就能取用。但如果您需要處理一些超過基本功能的事情， 例如建置華麗的使用者介面， 通常就需要安裝另一個套件來配合。

還好大多數第三方程式庫的安裝都相當簡單。程式庫的作者把它們打包成套件，「pip 套件管理器」可以用來安裝這些套件。

使用多個第三方套件需要一些技巧。有些套件需要先安裝其他套件來配合， 有些套件還可能與其他套件衝突，我們還可以根據需要安裝特定版本的套件。之前是否有提到電腦上的某些應用程式和作業系統元件依賴於特定的Python套件呢？這就是為什麼虛擬環境存在的原因。

虛擬環境就像沙盒， 您可以在其中只安裝特定於某個專案所需的Python套件，避免這些套件與其他專案（或您的系統）產生衝突的風險。您可以為每個專案建立不同的小沙盒， 只在其中安裝需要的套件， 讓一切都能保持整齊有序。實際上我們並沒有改變系統上已安裝的Python 套件， 因此可以避免破壞與您的專案無關但又很重要的事物。

#### 建立虛擬環境

每個虛擬環境都放在一個專屬的目錄中。通常我們會把這個資料夾命名為env或venv。對於每個專案， 我通常喜歡在專案資料夾內建立一個專屬的虛擬環境。Python提供了一個名為venv 的工具來完成這個工作。如果您使用Git或其他版本控制系統( version control system, VCS ) 來追蹤程
式碼的變更， 還有一個額外的設定步驟在後面會提到。

若想要在目前的工作目錄中建立名為venv 的虛擬環境， 請在命令列中執行以下指令：
```bash
python -m venv .venv
```
第一個venv是一個用來建立虛擬環境的指令，而第二個venv是指定虛擬環境的路徑。在上述的範例中，venv 只是個相對路徑，會在目前的工作目錄
中建立一個「.venv/ 」目錄。不過，也可以使用絕對路徑 並且可以任意取名。現在如果査看專案的工作目錄， 就會看到「venv/ 」資料夾已經被建立了。

若使用 Git 或 GitHub 進行版本控制，通常不會將 .venv 目錄加入版本控制，而是將其加入 .gitignore：
```bash
.venv/
```
這是因為虛擬環境中包含大量可重新建立的檔案，不需要上傳至 GitHub。相反地，通常會將套件資訊記錄在 requirements.txt 中。
```bash
pip freeze > requirements.txt
```

其他開發者便可利用：
```bash
pip install -r requirements.txt
```
重新建立相同的套件環境。

#### 啟用虐擬環境（CLI）

建立虛擬環境後，必須先啟用（Activate）才能使用其中的 Python 解譯器與已安裝的套件。

- 在 Linux、macOS 或 GitHub Codespaces 中，可使用下列指令啟用虛擬環境：
```bash
.venv\Scripts\activate.bat
```
- 在 Windows 的命令提示字元（Command Prompt）中，請使用：
```bash
.venv\Scripts\activate.bat
```
- 在 Windows PowerShell 中，請使用：
```bash
.venv\Scripts\Activate.ps1
```
若 PowerShell 出現執行政策（Execution Policy）相關錯誤，可在 PowerShell 中執行：
```
Set-ExecutionPolicy RemoteSigned -Scope CurrentUser
```

啟用成功後，命令提示字元前方通常會出現虛擬環境名稱，例如：
```bash
(.venv) student@codespaces:~/project$
```
其中`(.venv)`表示目前正在使用名為 .venv 的虛擬環境。此時執行：`python --version`或`pip list`都將使用虛擬環境中的 Python 與套件，而不會影響系統中的其他 Python 專案。當工作完成後，可使用下列指令離開虛擬環境：`deactivate`。執行後，命令提示字元前方的`(.venv)`將會消失，表示已回到系統預設的 Python 環境。

本課程主要使用 GitHub Codespaces，因此通常只需要記住： 
```bash
python -m venv .venv 

source .venv/bin/activate 

deactivate
```

#### 課堂練習

請回答：

1 為什麼不建議將 .venv 目錄上傳到 GitHub？

    因為 .venv 內含整個虛擬環境，包括 Python 執行檔和所有已安裝套件。
    缺點：

        檔案數量非常多，會讓 Git 儲存庫變得很大。
        不同作業系統（Windows、macOS、Linux）的 .venv 內容不同，可能無法在別人的電腦上使用。
        已經可以透過 requirements.txt 重新安裝套件，沒必要上傳 .venv。

    因此通常會把 .venv 加入 .gitignore：.venv/

2 為什麼建議將 requirements.txt 加入 Git 儲存庫？

    因為 requirements.txt 紀錄了專案需要的套件及版本。
    例如：
        flask==3.1.2
        requests==2.32.4
        openai==2.1.0

    其他人拿到專案後只要執行：
        pip install -r requirements.txt
    
    就能安裝完全相同的套件環境。

    這能確保：

        每個人使用相同版本
        減少環境設定時間
        方便部署到雲端平台

3 requirements.txt 如何協助團隊合作？
    假設你把專案放到 GitHub。

    組員只需要：
        git clone 專案網址
        python -m venv .venv
        pip install -r requirements.txt
    即可建立與你相同的環境。

    這樣大家：

        使用相同套件版本
        執行結果較一致
        不必逐一詢問要安裝哪些套件

    能大幅降低「我的電腦可以跑，你的電腦不能跑」的問題。    

4 如果兩位組員安裝了不同版本的 Flask，可能產生哪些問題？

    例如：
        甲同學：Flask 2.2
        乙同學：Flask 3.1
    可能發生：

        程式無法執行

        某些函式在新版或舊版已被修改或移除。

        錯誤訊息不同

        同一份程式可能在一台電腦正常，在另一台電腦出錯。

        功能行為不同

        新舊版本的預設設定可能不一樣，導致結果不同。

        除錯困難

        組員看到的問題彼此無法重現，增加溝通成本。

    因此通常會在 requirements.txt 指定版本：flask==3.1.2，讓所有組員安裝同一版本，避免版本衝突。

#### 課堂練習

- 為什麼團隊專案不一定適合隨時升級到最新版本？
    因為新版套件雖然可能修正錯誤或增加功能，但也可能：

        移除舊功能（Deprecated / Removed），造成程式報錯。
        與其他套件版本衝突。
        團隊成員的開發環境可能因此不一致。

        例如：
            # 原本 Flask 2.x 可以執行
            some_function()

            # Flask 3.x 已移除
            # 執行時可能出現錯誤
        
        因此團隊通常會先測試新版套件是否相容，再決定是否升級，而不是看到新版本就立刻更新。

- 為何需要將套件版本記錄在 requirements.txt 中？
    requirements.txt 的目的是：

        記錄專案使用的套件與版本。
        讓其他開發者能建立相同環境。
        避免因版本不同而產生錯誤。
        方便部署到伺服器或雲端。

    例如：
        Flask==3.1.0
        requests==2.32.3
        numpy==2.1.2

    其他人只要執行：
        pip install -r requirements.txt，就能安裝與專案相同的套件版本。

    如果沒有記錄版本：

        你的電腦可能是 Flask 3.1
        同學的電腦可能是 Flask 2.3
        兩人執行結果可能不同，增加除錯困難。 

- 更新套件後，為什麼要重新測試程式？

    因為升級後可能出現：
        1.相容性問題:舊程式碼無法配合新版本。
        2.新的 Bug:新版雖然修復舊問題，但可能產生新的問題。
        3.套件依賴改變:升級 Flask 時，相關套件版本也可能跟著變動。
        4.功能行為改變:程式能執行，但結果與以前不同。

    例如：pip install --upgrade flask

    升級後應測試：

        網頁是否能正常開啟
        表單是否正常送出
        資料庫是否正常連線
        API 是否正常回傳資料

    確認一切正常後，再更新：pip freeze > requirements.txt

    最後才提交到 GitHub：
        git add requirements.txt
        git commit -m "Update package versions"
        git push

#### 移除套件

若要移除已安裝的套件，可使用：

```bash
pip uninstall package_name
```
移除套件時要注意，當移除某個套件時：`pip uninstall flask`，通常只會移除 Flask 本身，而不一定會移除所有相依套件。原因是他套件可能也正在使用這些相依套件。

為什麼安裝一個套件時，可能會同時安裝許多其他套件？

    因為許多套件並不是獨立運作的，而是需要依賴其他套件提供功能，這些被依賴的套件稱為 Dependencies（相依套件）。

    ✅ 結論： 安裝套件時，pip 會自動安裝它所需要的相依套件，讓程式能正常運作。

為什麼移除套件時需要注意相依性（Dependencies）？

    因為某些套件可能被多個程式或套件共同使用。

    例如：
        Project A
        ↓
        requests

        Project B
         ↓
        requests
        ``
    如果直接移除：pip uninstall requests 可能會導致：

        Project A 無法執行
        Project B 也無法執行

    因此移除套件前要確認：

        是否還有其他專案需要它
        是否有其他套件依賴它

    ✅ 結論： 隨意移除套件可能造成其他程式故障，所以必須確認相依關係。

為什麼使用虛擬環境後，可以更放心地安裝與移除套件？

    虛擬環境（Virtual Environment）會建立一個獨立的 Python 套件空間。

    例如：
        電腦
        ├── 專案A (venv)
        │   └── Flask 2.3
        │
        └── 專案B (venv)
            └── Flask 3.1
    
    不同專案的套件彼此不影響。

    因此：

        安裝新套件較安全
        升級套件較安全
        移除套件較安全
        不會破壞其他專案

    即使把虛擬環境中的 Flask 刪掉：pip uninstall flask 也只影響目前的虛擬環境，不會影響電腦上其他 Python 專案。

    ✅ 結論： 虛擬環境把各專案隔離開來，大幅降低套件衝突與誤刪風險。

如果虛擬環境損壞或套件版本混亂，你會如何恢復開發環境？
   
    最簡單的方法是：
        
        步驟 1：刪除舊虛擬環境：
            
            Mac/Linux：rm -rf .venv
            Windows：rmdir /s .venv
        
        步驟 2：重新建立虛擬環境：python -m venv .venv

        步驟 3：啟動虛擬環境：

            Mac/Linux：source .venv/bin/activate
            Windows：.venv\Scripts\activate
                    `
        步驟 4：根據 requirements.txt 重新安裝：
            pip install -r requirements.txt
        
        步驟 5：確認安裝結果：pip list

[!WARNING]

不要隨意使用 sudo pip

在 Linux 或 macOS 系統中，除非非常清楚自己在做什麼，否則不建議使用：

sudo pip install package_name 這個指令可能將套件安裝到系統 Python 環境中， 並覆蓋或干擾作業系統原本的套件設定。

輕則造成 Python 環境混亂， 重則可能影響系統工具正常運作。

對於大多數 Python 專案而言， 使用虛擬環境（venv）即可解決套件安裝與版本管理問題。

建議流程如下：

- python -m venv .venv
- source .venv/bin/activate
- pip install package_name

如此套件只會安裝在目前專案的虛擬環境中， 不會影響系統中的其他 Python 專案。

在本課程使用的 GitHub Codespaces 環境中， 原則上不需要使用 sudo pip。
